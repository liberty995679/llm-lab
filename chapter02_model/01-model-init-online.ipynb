# 调用线上的大模型

## 1.调用具体模型厂商的API

### 1.1 调用deepseek官网的 API

调用大模型设计到三个特别重要的参数：
base_usl、model_id、api_key

举例1：将参数信息存储在.env配置文件中

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

from langchain_deepseek import ChatDeepSeek
from langchain_openai import ChatOpenAI

# 1.读取.env配置文件中的信息,相关的环境变量以.nev文件中为准
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 2.模型的初始化
llm_deepseek = ChatDeepSeek(
    model="deepseek-v4-pro",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL,
)

# 3.调用模型
response = llm_deepseek.invoke("请用一句话介绍你自己")
print(response)
```

## 3.使用中转平台
### 3.1 使用OpenRouter
### 举例1：调用的国内的大模型

```python
import os
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 加载配置文件
load_dotenv(override=True)

OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

model = ChatOpenRouter(
    model="deepseek/deepseek-v4-flash",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)

print(model.invoke("一句话介绍下你自己"))
```

举例2：调用的国外的大模型，例如GPT

```python
import os
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 加载配置文件
load_dotenv(override=True)

OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

model = ChatOpenAI(
    model="openai/gpt-4.1-mini",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
    max_tokens=1024, # 限制最大输出长度
)

print(model.invoke("一句话介绍下你自己"))
```

## 4.使用langchain1.X的统一使用方式

使用 inin_chat_model

举例1：调用Deepseek模型

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 加载配置文件
load_dotenv(override=True)

DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

print(model.invoke("中国GDP是多少"))
```
