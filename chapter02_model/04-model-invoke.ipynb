```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model = "deepseek-v4-flash",
    model_provider= "deepseek",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL,
)

message = [
    {"role": "system", "content": "你是一个专业的数学家"},
    {"role":"user","content":"我有6个苹果，分给两个人怎么分"},
]

response1 =  model.invoke(message)
print(f"AI的第一次回复：{response1.content}")

message.append({"role":"assistant","content":response1.content})

message1 = {"role": "user", "content": "那如果是三个人呢"}

message.append(message1)

response2 = model.invoke(message)

print(f"AI的第二次回复：{response2.content}")
```

### 1.3 消息列表的输入
举例

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model = "deepseek-v4-flash",
    model_provider= "deepseek",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL,
)

message = [
  SystemMessage(content="你是一个专业的数学家"),
    HumanMessage(content="我有6个苹果，分给两个人怎么分"),
]

response1 =  model.invoke(message)
print(f"AI的第一次回复：{response1.content}")

message.append(AIMessage(content=response1.content))
message.append(HumanMessage(content="那如果是三个人呢"))

response2 = model.invoke(message)

print(f"AI的第二次回复：{response2.content}")
```
